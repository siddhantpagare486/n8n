## 3. ***Data Wrangling***

### Data Wrangling Code

```python
# Write your code to make your dataset analysis ready.
# 1. Split the 'salary_est_in_$k' column at the hyphen (-) into temporary min and max columns
salary_split = data['salary_est_in_$k'].str.split('-', expand=True)

# 2. Convert the split string values into numeric format (floats/integers)
data['min_salary'] = pd.to_numeric(salary_split[0], errors='coerce')
data['max_salary'] = pd.to_numeric(salary_split[1], errors='coerce')

# 3. Multiply by 1,000 since the original scale is in thousands (e.g., '53' becomes 53,000)
data['min_salary'] = data['min_salary'] * 1000
data['max_salary'] = data['max_salary'] * 1000

# 4. Compute the average salary midpoint to use as your machine learning target variable
data['avg_salary'] = (data['min_salary'] + data['max_salary']) / 2

# Verify the new columns
print(data[['salary_est_in_$k', 'min_salary', 'max_salary', 'avg_salary']].head(10))

# Drop the original 'salary_est_in_$k' column since we have already extracted min_salary, max_salary, and avg_salary from it
data = data.drop(columns=['salary_est_in_$k'])
```

```python
# Clean the 'Company Name' column by splitting at the newline character ('\n')
# and keeping only the first part (the actual company name)
data['Company Name'] = data['Company Name'].astype(str).str.split('\n').str[0].str.strip()
```

```python
# Split the 'Location' column into 'job_city' and 'job_state' at the comma separator
location_split = data['Location'].astype(str).str.split(',', expand=True)

# Assign the city and state parts to new columns in your dataset
data['job_city'] = location_split[0].str.strip()
data['job_state'] = location_split[1].str.strip() if location_split.shape[1] > 1 else ''

# Verify the new location columns
print(data[['Location', 'job_city', 'job_state']].head(10))
# drop orignal col
data = data.drop(columns=['Location'])
```

```python
from sklearn.preprocessing import LabelEncoder

# Initialize the Label Encoder
le = LabelEncoder()

# Apply Label Encoding to categorical text columns
# This converts text categories (like employee counts or revenue ranges) into numerical codes (0, 1, 2, etc.)
for col in ['Size', 'Revenue', 'Type of ownership']:
    if col in data.columns:
        # Fill any missing values with a placeholder string first to avoid errors during encoding
        data[col] = data[col].fillna('Unknown')
        data[col + '_encoded'] = le.fit_transform(data[col])

# Verify the encoded columns
print(data[['Size', 'Size_encoded', 'Revenue', 'Revenue_encoded']].head(10))
```

```python
# Drop the original categorical text columns since they have been successfully label-encoded
cols_to_drop = ['Size', 'Revenue', 'Type of ownership']
data = data.drop(columns=cols_to_drop)

# Verify the updated columns list
print(data.columns.tolist())
```

```python
# Create binary flag columns by checking if specific tech keywords appear in the 'Job Description'
# We use .str.lower() to ensure the check is case-insensitive (e.g., matching 'Python' or 'python')

data['has_python'] = data['Job Description'].str.lower().str.contains('python', na=False).astype(int)
data['has_sql'] = data['Job Description'].str.lower().str.contains('sql', na=False).astype(int)
data['has_aws'] = data['Job Description'].str.lower().str.contains('aws', na=False).astype(int)
data['has_spark'] = data['Job Description'].str.lower().str.contains('spark', na=False).astype(int)
data['has_excel'] = data['Job Description'].str.lower().str.contains('excel', na=False).astype(int)

# Verify the newly created feature flags
print(data[['Job Description', 'has_python', 'has_sql', 'has_aws', 'has_spark', 'has_excel']].head(3))
```

## ***5. Hypothesis Testing***

### Based on your chart experiments, define three hypothetical statements from the dataset. In the next three questions, perform hypothesis testing to obtain final conclusion about the statements through your code and statistical testing.

### Hypothetical Statement - 1

#### 1. State Your research hypothesis as a null hypothesis and alternate hypothesis.

**Null Hypothesis ($H_0$):** There is no significant difference in the average salary between job postings that require Python (has_python = 1) and those that do not (has_python = 0).

**Alternative Hypothesis ($H_1$):** Job postings that require Python have a significantly higher average salary than those that do not.

#### 2. Perform an appropriate statistical test.

```python
# Perform Statistical Test to obtain P-Value

from scipy import stats

# Separate salary data into two groups: Python required vs. Not required
salary_with_python = data[data['has_python'] == 1]['avg_salary'].dropna()
salary_without_python = data[data['has_python'] == 0]['avg_salary'].dropna()

# Perform Independent Two-Sample t-test (Welch's t-test)
t_stat, p_val = stats.ttest_ind(salary_with_python, salary_without_python, equal_var=False)

print(f"T-statistic: {t_stat:.4f}")
print(f"P-value: {p_val}")

# Decision rule
alpha = 0.05
if p_val < alpha:
    print("Conclusion: Reject the null hypothesis (H0). There is a statistically significant difference in salaries.")
else:
    print("Conclusion: Fail to reject the null hypothesis (H0). No significant difference found.")
```

##### Which statistical test have you done to obtain P-Value?

An independent two-sample Welch's t-test (scipy.stats.ttest_ind with equal_var=False) was performed to obtain the P-value.

##### Why did you choose the specific statistical test?

Welch's t-test was chosen because we are comparing the means of a continuous numerical variable (avg_salary) across two independent categorical groups (jobs requiring Python vs. jobs not requiring Python). Setting equal_var=False is a robust statistical practice because it does not assume that the two groups share equal population variances.

What is the final conclusion from the statistical test Hypothetical Statement 1?

**Statistical Decision:** Since the resulting P-value ( 1.63×10−15 ) is exceptionally smaller than the standard significance level ( α=0.05 ), we reject the null hypothesis ( H0 ).

**Business Conclusion:** There is overwhelming statistical evidence to prove that job postings requiring Python offer a significantly higher average salary than those that do not. This confirms that Python proficiency carries a measurable and genuine market premium in the tech job dataset.

### Hypothetical Statement - 2

#### 1. State Your research hypothesis as a null hypothesis and alternate hypothesis.

**Null Hypothesis ($H_0$):** There is no correlation between a company's Glassdoor rating and the average salary it offers.

**Alternative Hypothesis ($H_1$):** There is a statistically significant correlation between company ratings and average salaries.

#### 2. Perform an appropriate statistical test.

```python
# Perform Statistical Test to obtain P-Value
from scipy import stats

# Drop rows where either rating or avg_salary is missing
rating_data = data[['Rating', 'avg_salary']].dropna()

# Perform Pearson correlation test
corr_coef, p_val_corr = stats.pearsonr(rating_data['Rating'], rating_data['avg_salary'])

print(f"Correlation Coefficient: {corr_coef:.4f}")
print(f"P-value: {p_val_corr}")

# Decision rule
alpha = 0.05
if p_val_corr < alpha:
    print("Conclusion: Reject the null hypothesis (H0). There is a statistically significant correlation.")
else:
    print("Conclusion: Fail to reject the null hypothesis (H0). No significant correlation found.")
```

##### Which statistical test have you done to obtain P-Value?

A Pearson correlation test (scipy.stats.pearsonr) was performed to obtain the P-value and correlation coefficient.

##### Why did you choose the specific statistical test?

The Pearson correlation test was chosen because both variables being evaluated—Rating (company rating) and avg_salary (average salary)—are continuous numerical features, and the goal was to measure the strength and direction of their linear relationship.

### Hypothetical Statement - 3

#### 1. State Your research hypothesis as a null hypothesis and alternate hypothesis.

**Null Hypothesis ($H_0$):** There is no significant difference in average salaries across different company revenue categories (Revenue_encoded).

**Alternative Hypothesis ($H_1$):** There is a statistically significant difference in average salaries across different company revenue categories.

#### 2. Perform an appropriate statistical test.

```python
# Perform Statistical Test to obtain P-Value
from scipy import stats

# Drop missing values for revenue encoding and average salary
revenue_data = data[['Revenue_encoded', 'avg_salary']].dropna()

# Extract salary arrays for each unique revenue category group
revenue_groups = [group['avg_salary'].values for name, group in revenue_data.groupby('Revenue_encoded')]

# Perform One-Way ANOVA test
f_stat, p_val_anova = stats.f_oneway(*revenue_groups)

print(f"F-statistic: {f_stat:.4f}")
print(f"P-value: {p_val_anova}")

# Decision rule
alpha = 0.05
if p_val_anova < alpha:
    print("Conclusion: Reject the null hypothesis (H0). There is a statistically significant difference in salaries across revenue tiers.")
else:
    print("Conclusion: Fail to reject the null hypothesis (H0). No significant difference found across revenue tiers.")
```

##### Which statistical test have you done to obtain P-Value?

A One-Way ANOVA (Analysis of Variance) test (scipy.stats.f_oneway) was performed to obtain the P-value.

##### Why did you choose the specific statistical test?

The One-Way ANOVA test was chosen because we needed to compare the means of a continuous numerical variable (avg_salary) across three or more independent categorical groups (the different company revenue brackets represented by Revenue_encoded). ANOVA is the standard parametric test used to determine whether there are any statistically significant differences between the means of three or more independent groups.

### 2. Handling Outliers

```python
# Handling Outliers & Outlier treatments
# Detect and handle extreme outliers in avg_salary using the Interquartile Range (IQR) method

Q1 = data['avg_salary'].quantile(0.25)
Q3 = data['avg_salary'].quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

# Filter the dataset to retain only records within the non-outlier threshold
data = data[(data['avg_salary'] >= lower_bound) & (data['avg_salary'] <= upper_bound)]

print(f"Dataset shape after handling outliers: {data.shape}")
```

##### What all outlier treatment techniques have you used and why did you use those techniques?

**Interquartile Range (IQR) Filtering (Trimming)**

```python
# Check what columns are currently present in your DataFrame
import re
import pandas as pd

# Recreate 'cleaned_job_description' from the existing 'Job Description' column
def simple_clean(text):
    if pd.isna(text):
        return ""
    text = str(text).lower()
    text = re.sub(r'http\S+|www\S+|https\S+', '', text, flags=re.MULTILINE) # remove URLs
    text = re.sub(r'[^\w\s]', '', text) # remove punctuation
    text = re.sub(r'\d+', '', text) # remove digits
    return text

data['cleaned_job_description'] = data['Job Description'].apply(simple_clean)
print("Successfully created 'cleaned_job_description'. Current columns count:", len(data.columns))
```

```python
# Remove Punctuations
import string

# Remove all punctuation marks
data['cleaned_job_description'] = data['cleaned_job_description'].apply(
    lambda x: str(x).translate(str.maketrans('', '', string.punctuation))
)
```

#### 4. Removing URLs & Removing words and digits contain digits.

```python
# Remove URLs & Remove words and digits contain digits
import re

def remove_urls_and_digits(text):
    # Remove URLs
    text = re.sub(r'http\S+|www\S+|https\S+', '', text, flags=re.MULTILINE)
    # Remove any words and numbers containing digits
    text = re.sub(r'\w*\d\w*', '', text)
    return text

data['cleaned_job_description'] = data['cleaned_job_description'].apply(remove_urls_and_digits)
```

#### 5. Removing Stopwords & Removing White spaces

```python
# Remove Stopwords and White spaces
import nltk
from nltk.corpus import stopwords

nltk.download('stopwords')
stop_words = set(stopwords.words('english'))

def remove_stopwords_and_whitespace(text):
    # Remove stopwords and excess whitespaces
    words = text.split()
    filtered_words = [w for w in words if w not in stop_words]
    return ' '.join(filtered_words)

data['cleaned_job_description'] = data['cleaned_job_description'].apply(remove_stopwords_and_whitespace)
```

### 5. Data Transformation

#### Do you think that your data needs to be transformed? If yes, which transformation have you used. Explain Why?

Yes. Real-world salary data typically exhibits a right-skewed distribution, where a large concentration of jobs falls within standard or median pay ranges, and a smaller long tail extends into significantly higher executive or specialized salary brackets.

**Which transformation have you used?**

A Logarithmic Transformation (specifically natural logarithm using np.log1p) was applied to the continuous target variable (avg_salary).

**Explain Why?**

* **Normalizes Distribution:** Salaries do not follow a normal (Gaussian) distribution natively. Log transformation compresses the right tail, pulling the skewed distribution closer to a normal bell-curve shape.

* **Stabilizes Variance:** It helps stabilize variance across different salary tiers (homoscedasticity), preventing high-income data points from exerting disproportionate leverage on model training.

* **Algorithm Compatibility:** Many parametric machine learning models (such as linear regression and neural networks) perform better, converge faster, and yield more reliable coefficients when input features and target variables approximate a normal distribution.

```python
# Transform Your data
import numpy as np
import pandas as pd

# Check skewness of the target variable avg_salary
skewness = data['avg_salary'].skew()
print(f"Skewness of avg_salary before transformation: {skewness:.4f}")

# Apply Log Transformation (log1p) to normalize right-skewed salary distributions
data['log_avg_salary'] = np.log1p(data['avg_salary'])

print("Log transformation applied successfully. Created 'log_avg_salary'.")
```
